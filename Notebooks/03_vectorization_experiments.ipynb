# Bag of Word (BOW)

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
documents = [
    "win free prize",
    "free prize today",
    "meeting today"
]

vectorizer = CountVectorizer()

X = vectorizer.fit_transform(documents)
```

```python
print(vectorizer.get_feature_names_out())
```

```python
print(X.toarray())
```

```python
documents = [
    "free free free prize",
    "free prize today",
    "meeting today"
]

vectorizer = CountVectorizer()

X = vectorizer.fit_transform(documents)

print(vectorizer.get_feature_names_out())
print(X.toarray())
```

```python
texts = df["Text"].fillna("")
y = df["Spam"]
```

```python
bow_vectorizer = CountVectorizer()

X_bow = bow_vectorizer.fit_transform(texts)
```

```python
print("Shape:", X_bow.shape)
```

```python
print("Number of vocabulary words:",
      len(bow_vectorizer.get_feature_names_out()))
```

```python
print("Matrix type:", type(X_bow))
```

# Binary BoW

```python
binary_vectorizer = CountVectorizer(binary=True)

X_binary = binary_vectorizer.fit_transform(texts)
```

```python
print("Shape:", X_binary.shape)

print(
    "Vocabulary size:",
    len(binary_vectorizer.get_feature_names_out())
)

print("Matrix type:", type(X_binary))
```

```python
test_documents = [
    "free free free prize",
    "free prize today",
    "meeting today"
]
```

```python
count_vectorizer = CountVectorizer()

X_count = count_vectorizer.fit_transform(test_documents)

print("Vocabulary:")
print(count_vectorizer.get_feature_names_out())

print("\nNormal BoW:")
print(X_count.toarray())
```

```python
binary_vectorizer_test = CountVectorizer(binary=True)

X_binary_test = binary_vectorizer_test.fit_transform(test_documents)

print("\nBinary BoW:")
print(X_binary_test.toarray())
```
