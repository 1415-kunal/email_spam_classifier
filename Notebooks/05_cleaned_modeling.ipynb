```python
import pandas as pd

from src.preprocessing import (
    tokenize_text,
    stem_tokens,
    lowercase_text,
    replace_urls,
    replace_email_addresses,
    remove_numbers,
    remove_punctuation,
    remove_stopwords,
    lemmatize_text
)
```

```python
df = pd.read_csv("../data/emails.csv")

print("Shape:", df.shape)
print(df.head())
```

```python
X = df["Text"].fillna("")
y = df["Spam"]
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Train:", X_train.shape)
print("Test :", X_test.shape)
```

```python
def clean_text_a(text):
    text = lowercase_text(text)
    text = replace_urls(text)
    text = replace_email_addresses(text)
    text = remove_numbers(text)
    text = remove_punctuation(text)
    text = remove_stopwords(text)
    return text
```

```python
X_train_clean_a = X_train.apply(clean_text_a)
X_test_clean_a = X_test.apply(clean_text_a)
```

```python
for i in range(3):
    print("ORIGINAL:")
    print(X_train.iloc[i])

    print("\nCLEANED:")
    print(X_train_clean_a.iloc[i])

    print("\n" + "=" * 80 + "\n")
```

```python
sample = """
Congratulations! Visit http : / / www . example . com
or contact test @ example . com today! You won 1000 dollars!!!
"""

print("Original:")
print(sample)

print("\nCleaned:")
print(clean_text_a(sample))
```

```python
print("Training samples:", len(X_train_clean_a))
print("Testing samples :", len(X_test_clean_a))

print("Missing train:", X_train_clean_a.isna().sum())
print("Missing test :", X_test_clean_a.isna().sum())
```

```python
def clean_text_b(text):
    text = clean_text_a(text)

    tokens = tokenize_text(text)
    tokens = stem_tokens(tokens)

    return " ".join(tokens)
```

```python
X_train_clean_b = X_train.apply(clean_text_b)
X_test_clean_b = X_test.apply(clean_text_b)
```

```python
for i in range(3):
    print("ORIGINAL:")
    print(X_train.iloc[i])

    print("\nCLEAN A:")
    print(X_train_clean_a.iloc[i])

    print("\nCLEAN B - STEMMED:")
    print(X_train_clean_b.iloc[i])

    print("\n" + "="*80 + "\n")
```

```python
print("Training samples:", len(X_train_clean_b))
print("Testing samples :", len(X_test_clean_b))

print("Missing train:", X_train_clean_b.isna().sum())
print("Missing test :", X_test_clean_b.isna().sum())
```

```python
def clean_text_c(text):
    text = clean_text_a(text)

    tokens = lemmatize_text(text)

    return " ".join(tokens)
```

```python
X_train_clean_c = X_train.apply(clean_text_c)
X_test_clean_c = X_test.apply(clean_text_c)
```

```python
for i in range(3):
    print("ORIGINAL:")
    print(X_train.iloc[i])

    print("\nCLEAN A:")
    print(X_train_clean_a.iloc[i])

    print("\nCLEAN B - STEMMING:")
    print(X_train_clean_b.iloc[i])

    print("\nCLEAN C - LEMMATIZATION:")
    print(X_train_clean_c.iloc[i])

    print("\n" + "="*80 + "\n")
```

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
```

```python
def create_bow_features(X_train, X_test):
    vectorizer = CountVectorizer()

    X_train_vec = vectorizer.fit_transform(X_train)
    X_test_vec = vectorizer.transform(X_test)

    return X_train_vec, X_test_vec, vectorizer
```

```python
def create_binary_bow_features(X_train, X_test):
    vectorizer = CountVectorizer(binary=True)

    X_train_vec = vectorizer.fit_transform(X_train)
    X_test_vec = vectorizer.transform(X_test)

    return X_train_vec, X_test_vec, vectorizer
```

```python
def create_tfidf_features(X_train, X_test):
    vectorizer = TfidfVectorizer()

    X_train_vec = vectorizer.fit_transform(X_train)
    X_test_vec = vectorizer.transform(X_test)

    return X_train_vec, X_test_vec, vectorizer
```

```python
X_train_a_bow, X_test_a_bow, bow_a_vectorizer = create_bow_features(
    X_train_clean_a,
    X_test_clean_a
)

print("BoW")
print("Train:", X_train_a_bow.shape)
print("Test :", X_test_a_bow.shape)
print("Vocab:", len(bow_a_vectorizer.vocabulary_))
```

```python
X_train_a_binary, X_test_a_binary, binary_a_vectorizer = create_binary_bow_features(
    X_train_clean_a,
    X_test_clean_a
)

print("Binary BoW")
print("Train:", X_train_a_binary.shape)
print("Test :", X_test_a_binary.shape)
print("Vocab:", len(binary_a_vectorizer.vocabulary_))
```

```python
X_train_a_tfidf, X_test_a_tfidf, tfidf_a_vectorizer = create_tfidf_features(
    X_train_clean_a,
    X_test_clean_a
)

print("TF-IDF")
print("Train:", X_train_a_tfidf.shape)
print("Test :", X_test_a_tfidf.shape)
print("Vocab:", len(tfidf_a_vectorizer.vocabulary_))
```

```python
X_train_b_bow, X_test_b_bow, bow_b_vectorizer = create_bow_features(
    X_train_clean_b,
    X_test_clean_b
)

X_train_b_binary, X_test_b_binary, binary_b_vectorizer = create_binary_bow_features(
    X_train_clean_b,
    X_test_clean_b
)

X_train_b_tfidf, X_test_b_tfidf, tfidf_b_vectorizer = create_tfidf_features(
    X_train_clean_b,
    X_test_clean_b
)
```

```python
print("Clean B - BoW     :", X_train_b_bow.shape)
print("Clean B - Binary  :", X_train_b_binary.shape)
print("Clean B - TF-IDF  :", X_train_b_tfidf.shape)
```

```python
X_train_c_bow, X_test_c_bow, bow_c_vectorizer = create_bow_features(
    X_train_clean_c,
    X_test_clean_c
)

X_train_c_binary, X_test_c_binary, binary_c_vectorizer = create_binary_bow_features(
    X_train_clean_c,
    X_test_clean_c
)

X_train_c_tfidf, X_test_c_tfidf, tfidf_c_vectorizer = create_tfidf_features(
    X_train_clean_c,
    X_test_clean_c
)
```

```python
print("Clean C - BoW     :", X_train_c_bow.shape)
print("Clean C - Binary  :", X_train_c_binary.shape)
print("Clean C - TF-IDF  :", X_train_c_tfidf.shape)
```

```python
def create_word_ngram_features(X_train, X_test, ngram_range=(1, 2)):
    vectorizer = TfidfVectorizer(
        ngram_range=ngram_range
    )

    X_train_vec = vectorizer.fit_transform(X_train)
    X_test_vec = vectorizer.transform(X_test)

    return X_train_vec, X_test_vec, vectorizer
```

```python
X_train_a_ngram, X_test_a_ngram, ngram_a_vectorizer = create_word_ngram_features(
    X_train_clean_a,
    X_test_clean_a,
    ngram_range=(1, 2)
)

print("Clean A - 1 to 2 grams")
print("Train:", X_train_a_ngram.shape)
print("Test :", X_test_a_ngram.shape)
print("Vocab:", len(ngram_a_vectorizer.vocabulary_))
```

```python
X_train_a_ngram3, X_test_a_ngram3, ngram3_a_vectorizer = create_word_ngram_features(
    X_train_clean_a,
    X_test_clean_a,
    ngram_range=(1, 3)
)

print("Clean A - 1 to 3 grams")
print("Train:", X_train_a_ngram3.shape)
print("Test :", X_test_a_ngram3.shape)
print("Vocab:", len(ngram3_a_vectorizer.vocabulary_))
```

# Clean A + BoW + MultinomialNB

```python
from sklearn.naive_bayes import MultinomialNB

nb_model = MultinomialNB()

nb_model.fit(X_train_a_bow, y_train)

y_pred_a_bow_nb = nb_model.predict(X_test_a_bow)
```

```python

accuracy = accuracy_score(y_test, y_pred_a_bow_nb)
precision = precision_score(y_test, y_pred_a_bow_nb)
recall = recall_score(y_test, y_pred_a_bow_nb)
f1 = f1_score(y_test, y_pred_a_bow_nb)

print("Clean A + BoW + MultinomialNB")
print("Accuracy :", accuracy)
print("Precision:", precision)
print("Recall   :", recall)
print("F1 Score :", f1)
```

```python
print(classification_report(
    y_test,
    y_pred_a_bow_nb,
    target_names=["HAM", "SPAM"]
))
```

```python
results = []

results.append({
    "Preprocessing": "Clean A",
    "Vectorizer": "BoW",
    "Model": "MultinomialNB",
    "Accuracy": accuracy,
    "Precision": precision,
    "Recall": recall,
    "F1": f1
})

results_df = pd.DataFrame(results)

results_df
```
