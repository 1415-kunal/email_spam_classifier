```python
import pandas as pd

from src.preprocessing import (
    tokenize_text,
    stem_tokens,
    lowercase_text,
    replace_urls,
    replace_email_addresses,
    remove_numbers,
    remove_punctuation,
    remove_stopwords,
    lemmatize_text
)
```

```python
df = pd.read_csv("../data/emails.csv")

print("Shape:", df.shape)
print(df.head())
```

```python
X = df["Text"].fillna("")
y = df["Spam"]
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Train:", X_train.shape)
print("Test :", X_test.shape)
```

```python
def clean_text_a(text):
    text = lowercase_text(text)
    text = replace_urls(text)
    text = replace_email_addresses(text)
    text = remove_numbers(text)
    text = remove_punctuation(text)
    text = remove_stopwords(text)
    return text
```

```python
X_train_clean_a = X_train.apply(clean_text_a)
X_test_clean_a = X_test.apply(clean_text_a)
```

```python
for i in range(3):
    print("ORIGINAL:")
    print(X_train.iloc[i])

    print("\nCLEANED:")
    print(X_train_clean_a.iloc[i])

    print("\n" + "=" * 80 + "\n")
```

```python
sample = """
Congratulations! Visit http : / / www . example . com
or contact test @ example . com today! You won 1000 dollars!!!
"""

print("Original:")
print(sample)

print("\nCleaned:")
print(clean_text_a(sample))
```

```python
print("Training samples:", len(X_train_clean_a))
print("Testing samples :", len(X_test_clean_a))

print("Missing train:", X_train_clean_a.isna().sum())
print("Missing test :", X_test_clean_a.isna().sum())
```

```python
def clean_text_b(text):
    text = clean_text_a(text)

    tokens = tokenize_text(text)
    tokens = stem_tokens(tokens)

    return " ".join(tokens)
```

```python
X_train_clean_b = X_train.apply(clean_text_b)
X_test_clean_b = X_test.apply(clean_text_b)
```

```python
for i in range(3):
    print("ORIGINAL:")
    print(X_train.iloc[i])

    print("\nCLEAN A:")
    print(X_train_clean_a.iloc[i])

    print("\nCLEAN B - STEMMED:")
    print(X_train_clean_b.iloc[i])

    print("\n" + "="*80 + "\n")
```

```python
print("Training samples:", len(X_train_clean_b))
print("Testing samples :", len(X_test_clean_b))

print("Missing train:", X_train_clean_b.isna().sum())
print("Missing test :", X_test_clean_b.isna().sum())
```

```python
def clean_text_c(text):
    text = clean_text_a(text)

    tokens = lemmatize_text(text)

    return " ".join(tokens)
```

```python
X_train_clean_c = X_train.apply(clean_text_c)
X_test_clean_c = X_test.apply(clean_text_c)
```

```python
for i in range(3):
    print("ORIGINAL:")
    print(X_train.iloc[i])

    print("\nCLEAN A:")
    print(X_train_clean_a.iloc[i])

    print("\nCLEAN B - STEMMING:")
    print(X_train_clean_b.iloc[i])

    print("\nCLEAN C - LEMMATIZATION:")
    print(X_train_clean_c.iloc[i])

    print("\n" + "="*80 + "\n")
```
