# remove urls

```python
def remove_urls(text):
    return re.sub(r'https?://\S+|www\.\S+', '', text)
```

```python
text = "Congratulations! Visit https://example.com/claim now!"

print("Original:")
print(text)

print("\nURLs removed:")
print(remove_urls(text))
```

```python
def replace_urls(text):
        return re.sub(
            r'https?://\S+|www\.\S+',
            'URL_TOKEN',
            text
        )
```

```python
text = "Congratulations! Visit https://example.com/claim now!"

print(replace_urls(text))
```

```python
text = "FREE prize! Click https://example.com/win now!"

print("ORIGINAL:")
print(text)

print("\nURL REMOVED:")
print(remove_urls(text))

print("\nURL REPLACED:")
print(replace_urls(text))
```

# Stopword Handling

```python
import nltk
```

```python
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords
```

```python
stop_words=set(stopwords.words('english'))
```

```python
print("Number of stopwords:", len(stop_words))
```

```python
print(list(stop_words)[:20])
```

```python
def remove_stopwords(text):
    words = text.split()
    
    filtered_words = [
        word for word in words
        if word.lower() not in stop_words
    ]
    
    return ' '.join(filtered_words)
```

```python
text = "This is a very good offer for you"

print("Original:")
print(text)

print("\nAfter stopword removal:")
print(remove_stopwords(text))
```

```python
sample_text = df['Text'].iloc[0]

print("ORIGINAL:\n")
print(sample_text)

print("\nWITHOUT STOPWORDS:\n")
print(remove_stopwords(sample_text))
```

```python
def count_stopwords(text):
    words = text.split()

    return sum(
        1 for word in words
        if word.lower() in stop_words
    )
```

```python
df['stopword_count'] = df['Text'].apply(count_stopwords)
```

```python
df.groupby('Spam')['stopword_count'].describe()
```

```python
df['word_count'] = df['Text'].apply(lambda x: len(x.split()))

df['stopword_ratio'] = (
    df['stopword_count'] / df['word_count']
)
```

```python
df.groupby('Spam')['stopword_ratio'].describe()
```

# Stemming

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()
```

```python
def stem_tokens(tokens):
    return [stemmer.stem(token) for token in tokens]
```

```python
words = [
    "connect",
    "connected",
    "connecting",
    "connection",
    "study",
    "studies",
    "studying",
    "playing",
    "played"
]

stemmed_words = stem_tokens(words)

for original, stemmed in zip(words, stemmed_words):
    print(f"{original:15} -> {stemmed}")
```

```python
from nltk.tokenize import word_tokenize

sample = df['Text'].iloc[0]

tokens = word_tokenize(sample)
stemmed_tokens = stem_tokens(tokens)

print("Original tokens:")
print(tokens[:50])

print("\nStemmed tokens:")
print(stemmed_tokens[:50])
```

# Lemmatization

```python
nlp = spacy.load("en_core_web_sm")
```

```python
def lemmatize_text(text):
    doc = nlp(text)
    return [token.lemma_ for token in doc]
```

```python
words = [
    "connect",
    "connected",
    "connecting",
    "connection",
    "study",
    "studies",
    "studying",
    "running",
    "better"
]

text = " ".join(words)

lemmas = lemmatize_text(text)

for original, lemma in zip(words, lemmas):
    print(f"{original:15} -> {lemma}")
```

```python
sample = df['Text'].iloc[0]

doc = nlp(sample)

print("Original tokens:")
print([token.text for token in doc][:50])

print("\nLemmatized tokens:")
print([token.lemma_ for token in doc][:50])
```

# URLs and Email Addresses

```python
url_pattern = r'https?\s*:\s*/\s*/\s*(?:www\s*\.\s*)?[\w.-]+(?:\s*\.\s*[\w.-]+)+|www\s*\.\s*[\w.-]+(?:\s*\.\s*[\w.-]+)+'


def replace_urls(text):
    return re.sub(url_pattern, 'URL_TOKEN', text)
```

```python
sample = df['Text'].iloc[0]

print("Original:")
print(sample)

print("\nURL replaced:")
print(replace_urls(sample))
```

```python
email_pattern = r'\b[\w.-]+\s*@\s*[\w.-]+(?:\s*\.\s*[\w.-]+)+\b'

def replace_email_addresses(text):
    return re.sub(email_pattern, 'EMAIL_TOKEN', text)
```

```python
sample = df['Text'].iloc[10]

print("Original:")
print(sample)

print("\nEmail addresses replaced:")
print(replace_email_addresses(sample))
```

```python
def replace_urls_and_emails(text):
    text = replace_urls(text)
    text = replace_email_addresses(text)
    return text
```

```python
sample = df['Text'].iloc[10]
print("Original:")
print(sample)
print("\n")
print(replace_urls_and_emails(sample))
```
